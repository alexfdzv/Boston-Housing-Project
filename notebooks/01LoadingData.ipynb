# <span style="color:red">Boston Housing Projects - Loading Data</span>

Carga y valida el dataset crudo de Boston Housing.
 
Justificación:
    El primer paso de cualquier pipeline es cargar los datos de forma
    reproducible y verificar que la estructura sea la esperada.
    Esto garantiza que etapas posteriores reciban datos válidos.

## <span style="color:green">Configuración</span>

Importación de bibliotecas:

```python
import pandas as pd
import numpy as np
import yaml
import sys
import os
```

Extracción de columnas desde params.yaml

```python
with open('params.yaml', 'r') as f:
    params = yaml.safe_load(f)

columnas = params['data']['columns']
print("Columnas definidas en params.yaml:")
print(columnas)
```

El dataset Boston Housing proviene del UCI Machine Learning Repository.
Está separado por espacios múltiples (no por comas), por lo que se usa
sep=r'\s+' para parsearlo correctamente.

```python
ruta = params['data']['raw_path']

df = pd.read_csv(ruta, header=None, sep=r'\s+', names=columnas)

print(f"  Archivo cargado: {ruta}")
print(f"  Filas:    {df.shape[0]}")
print(f"  Columnas: {df.shape[1]}")
```
